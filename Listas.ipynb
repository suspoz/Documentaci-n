Creacion de Listas

```python
# Lista vacía
vacia = []
vacia_alt = list()
print(vacia_alt)
```

```python
# Lista con elementos
frutas = ["manzana", "pera", "naranja"]
mixta = [1, "dos", 3.0, True]
print(frutas,mixta)
```

```python
# Desde otro iterable
letras = list("hola") # ['h', 'o', 'l', 'a']
numeros = list(range(1,13)) # [0, 1, 2, 3, 4]
print(letras,numeros)
```

```python
# 2.Orden indexación (incluyendo índices negativos y slicing)

frutas = ["manzana", "pera", "naranja", "uva"]
print("Primero:", frutas[0]) # 'manzana'
print("Último:", frutas[-1]) # 'uva'
print("Penúltimo:", frutas[-2]) # 'naranja'
print("Slicing [1:3]:", frutas[1:3]) # ['pera', 'naranja']
```

```python
# Duplicados permitidos
notas = [10, 8, 10, 9, 8]
print("Con duplicados:", notas)

# Listas anidadas (Matriz 3x3)
matriz = [
[1, 2, 3],# fila 0,
[4, 5, 6],# fila 1,
[7, 8, 9]# fila 2,
]
print("Elemento fila 1, col 2:", matriz[1][2]) # 6
```

```python
#Mutabilidad (modificar elementos o rangos)Python
colores = ["rojo", "verde", "azul"]
colores[1] = "amarillo"
print("Un elemento modificado:", colores) # ['rojo', 'amarillo', 'azul']
colores[0:2] = ["negro", "blanco"]
print("Rango modificado:", colores) # ['negro', 'blanco', 'azul']
```

```python
#Dinamismo (añadir y quitar elementos en ejecución)Python
tareas = ["estudiar", "practicar"]
tareas.append("exponer") # Añade al final
print("Tras:", tareas)
tareas.insert(1, "repasar") # Inserta en posición 1
print("Tras añadir:", tareas)
eliminada = tareas.pop() # Quita y devuelve el último
print("Tras quitar:", eliminada)
print("Tras ultimo proceso:", tareas)
tareas.remove("repasar") # Quita por valor
print("Tras quitar:", tareas, "| Eliminada:", eliminada)
```

```python
#Iterabilidad (for, enumerate y list comprehension)Python
#frutas = ["manzana", "pera", "naranja"]

print("--- Bucle for directo ---")
for indice in frutas:
   print(indice)

cuadrados = [x ** 2 for x in range(5)]
print(cuadrados) # [0, 1, 4, 9, 16]
```

```python
#Principales métodos Python
nums = [3, 1, 2, 2, 4]
print(nums)
nums.append(5) # [3, 1, 2, 2, 4, 5]
print(nums)
nums.extend([6, 7]) # [3, 1, 2, 2, 4, 5, 6, 7]
print(nums)
nums.insert(1, 12) # [0, 3, 1, 2, 2, 4, 5, 6, 7], primero posicion y luego el valor
print(nums)
nums.remove(12) # Elimina el primer 12
print(nums)
ultimo = nums.pop() # Quita y devuelve el último (7)
print(nums,ultimo)
print("Índice del 4:", nums.index(4))
print("Veces que aparece el 2:", nums.count(2))
nums.sort(reverse=True)# Ordena de mayor a menor in-place
print("Ordenada desc:", nums)
nums.reverse() # Invierte el orden in-place
print("Invertida:", nums)
copia = nums.copy() # Copia superficial
print("Copia:", copia)
nums.clear() # Vacía la lista
print("Tras clear():", nums)
```
